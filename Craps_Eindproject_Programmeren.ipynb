### __Inzetregels Craps__
Er kan tegen en voor de shooter worden ingezet, (de "Pass line" en de "Dont Pass line") als er tegen de shooter wordt ingezet dan zullen diegenen winnen als de shooter verliest, word er voor de shooter ingezet dan winnen diegenen als de shooter wint. <br/>
<br/>
__voor de eerste gooi:__ <br/>
Een shooter verliest als de shooter 2,3 of 12 gooit, dus dan winnen degenen die tegen de shooter hebben ingezet, dit komt wel met zijn regels, in meeste casino's wordt er niet uitbetaald als de shooter 12 gooit, omdat de shooter een nadeel heeft met zijn uitwerpgooi. Deze regel wordt dus in de simulatie ook gehanteerd. <br/>
Een shooter wint als de shooter 7 of 11 gooit, dus hiermee winnen degenen die voor de shooter hebben ingezet, hier zitten geen extra regels aan vast. <br/> voor beiden scenario's geldt dat de inzet 1 keer wordt uitbetaald.<br/>
<br/>
__voor de rest van de ronde:__ <br/>
Hier wordt het ingewikkelder, de shooter die geen 2,3,7,11 of 12 heeft gegooid moet verder spelen, de uitwerpgooi die de shooter heeft gemaakt wordt zijn "point", dus als de shooter een uitwerpgooi heeft gemaakt van 8, dan is de shooter zijn point 8. <br/>
De shooter blijft hierna doorgooien totdat diegene of hetzelfde point weer gooit of een 7 gooit, als de shooter hetzelfde point weer gooit, dan wint de shooter en winnen degenen die voor de shooter hebben ingezet, als de shooter een 7 gooit dan verliest de shooter en winnen de mensen die tegen de shooter hebben ingezet, de rest van de inzetten worden van het bord geveegd. <br/>
de mensen die voor de shooter hebben ingezet, kunnen nadat de shooter een point heeft gegooid, extra inzetten plaatsen. Deze inzetten kunnen geplaatst worden op alle getallen, als het point bijvoorbeeld een 6 is en er wordt ingezet op 8, dan zet je in op de kans dat de nummer 8 wordt gegooid voordat er 6 of 7 wordt gegooid. Elke keer als er 8 wordt gegooid wint degene die heeft ingezet op de nummer 8, dit is onwaarschijnlijk, maar als het nummer 8 drie keer wordt gegooid voordat het spel eindigt, dan heeft degene die heeft ingezet op nummer 8, drie keer gewonnen. Dit geldt voor alle getallen en elk getal wordt anders uitbetaald. <br/>
6 of 8 = 6 tegen 5 uitbetaald, dit betekend dat er 1.2 x de inzet wordt betaald (10 euro ingezet, 12 euro winst) <br/>
5 of 9 = 3 tegen 2 uitbetaald, dit betekend dat er 1.5 x de inzet wordt betaald (10 euro ingezet, 15 euro winst) <br/>
4 of 10 = 2 tegen 1 uitbetaald, dit betekend dat er 2 x de inzet wordt betaald (10 euro ingezet, 20 euro winst) <br/>
<br/>
op een 2,3,11 of 12 kan alleen worden ingezet met een speciale eenworp-inzet, dit betekend dat je inzet op dat de volgende worp een 2,3,11 of 12 wordt. <br/> Deze payouts zijn als volgt: <br/>
3 of 11 = 15 tegen 1, dit betekend dat 15 x de inzet wordt betaald (10 euro ingezet, 150 euro winst) <br/>
2 of 12 = 30 tegen 1, dit betekend dat 30 x de inzet wordt betaald (10 euro ingezet, 300 euro winst) <br/>
<br/>
__*excentrieke inzetten:__ <br/>
__hardways:__ <br/>
dit zijn de inzetten die spelers kunnen doen opdat er dubbel wordt gegooid, ofwel hard 4, 6, 8 of 10, de payouts zijn als volgt: <br/>
hard 4 of hard 10: 7 tegen 1, ofwel 7 x de inzet wordt betaald (10 euro ingezet, 70 euro winst) <br/>
hard 6 of hard 8: 9 tegen 1, ofwel 9 x de inzet wordt betaald (10 euro ingezet, 90 euro winst) <br/>
__field bets:__ <br/>
een field bet is eenworp-inzet hiermee kan er gegokt worden dat er met de volgende worp een 2,3,4,9,10,11 en 12 wordt gegooid, als er een 2 wordt gegooid dan krijg je het dubbele van je inzet terug, als er een 12 wordt gegooid krijg je het driedubbele, als er een 3,4,9,10 of 11 wordt gegooid, dan krijg je je inzet terug. <br/>
__any craps:__ <br/>
hiermee wordt er ingezet dat er bij de volgende worp een 2,3 of 12 wordt gegooid, als dit gebeurd krijgt diegene 7 x zijn inzet, dus bij 10 euro krijgt diegene als die wint 70 euro terug. <br/>
__any seven__ <br/>
hiermee wordt er ingezet dat er bij de volgende worp een 7 wordt gegooid, als dit gebeurd krijgt diegene 4 x zijn inzet, dus bij 10 euro krijgt diegene als die wint 40 euro terug. <br/>
<br/>
<br/>
*-(de excentrieke inzetten worden apart behandeld in het strategie-verslag)
